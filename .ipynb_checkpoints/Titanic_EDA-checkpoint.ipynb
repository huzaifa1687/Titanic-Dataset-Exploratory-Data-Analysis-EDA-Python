### Dataset Structure and Initial Findings

The Titanic training dataset contains 891 passenger records and 12 columns.

The dataset includes numerical and categorical features. Numerical features include passenger class, age, fare, and family-related information. Categorical features include passenger name, sex, ticket, cabin, and port of embarkation.

The initial inspection identified missing values in three columns:

- Age: 177 missing values (19.87%).
- Cabin: 687 missing values (77.10%).
- Embarked: 2 missing values (0.22%).

The Cabin column has the highest percentage of missing values, while Age also requires attention. These missing values will be investigated before selecting suitable cleaning methods.

### Missing Values and Cleaning Strategy

The dataset contains missing values in Age, Cabin, and Embarked.

The Age column has 177 missing values (19.87%). Its mean is 29.70, median is 28, and skewness is 0.389. Since the median is less sensitive to extreme values, missing ages will be replaced with the median.

The Embarked column has only two missing values. Since Southampton (S) is the most frequent category, these values will be replaced with the mode.

The Cabin column has 687 missing values (77.10%). Instead of attempting unreliable cabin-number imputation, a new HasCabin feature will indicate whether cabin information was recorded. The original Cabin column will then be removed.

The dataset contains zero duplicate rows, so no duplicate records need to be removed.

```python
# Keep the original dataset unchanged
df_clean = df.copy()

# Create a feature indicating whether cabin information was recorded
df_clean["HasCabin"] = df_clean["Cabin"].notnull().astype(int)

# Fill missing Age values with the median
age_median = df_clean["Age"].median()
df_clean["Age"] = df_clean["Age"].fillna(age_median)

# Fill missing Embarked values with the mode
embarked_mode = df_clean["Embarked"].mode()[0]
df_clean["Embarked"] = df_clean["Embarked"].fillna(embarked_mode)

# Remove the original Cabin column
df_clean = df_clean.drop(columns=["Cabin"])

# Remove duplicate rows, if any
df_clean = df_clean.drop_duplicates()

print("Cleaning completed successfully!")
print("Cleaned dataset shape:", df_clean.shape)
```

```python
print("Missing values after cleaning:")
print(df_clean.isnull().sum())

print("\nTotal missing values:", df_clean.isnull().sum().sum())

print("\nDuplicate rows:", df_clean.duplicated().sum())
```

### Data Cleaning Results

The following cleaning operations were performed:

1. Missing Age values were replaced with the median age of 28.
2. Missing Embarked values were replaced with the most frequent category, Southampton (S).
3. A new HasCabin feature was created to indicate whether cabin information was available.
4. The original Cabin column was removed because approximately 77.10% of its values were missing.
5. Duplicate records were checked and none were found.

After cleaning, the dataset contains 891 rows and 12 columns, with zero missing values and zero duplicate records.

```python
# Select numerical columns for outlier detection
numeric_columns = ["Age", "Fare", "SibSp", "Parch"]

for col in numeric_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ]

    print(f"\nColumn: {col}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

### Outlier Detection Using the IQR Method

The Interquartile Range (IQR) method was used to identify potential outliers in Age, Fare, SibSp, and Parch.

The analysis identified:

- Age: 66 potential outliers.
- Fare: 116 potential outliers.
- SibSp: 46 potential outliers.
- Parch: 213 potential outliers.

The Parch feature has an IQR of zero because at least 75% of passengers have a value of zero. Therefore, all positive values are flagged as outliers by this method.

These observations are not necessarily errors. Unusually high fares, older passengers, and large family groups can represent genuine passenger records.

Therefore, the identified outliers will be retained rather than automatically removed. This preserves potentially meaningful information in the dataset.

```python
numeric_columns = ["Age", "Fare", "SibSp", "Parch"]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for col, ax in zip(numeric_columns, axes.flatten()):
    sns.boxplot(data=df_clean, y=col, ax=ax)
    ax.set_title(f"Boxplot of {col}")
    ax.set_ylabel(col)

plt.tight_layout()
plt.savefig("outlier_boxplots.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Boxplot Analysis and Outlier Interpretation

Boxplots were created to examine the distributions and potential outliers in Age, Fare, SibSp, and Parch.

**Age:** The median age is approximately 28 years. Most passengers fall between approximately 22 and 35 years. Several older passengers appear as potential outliers.

**Fare:** The distribution is strongly right-skewed. Most passengers paid relatively low fares, while some paid considerably higher amounts, with the maximum reaching approximately 512.

**SibSp:** Most passengers travelled without siblings or spouses. A small number of passengers had larger family groups.

**Parch:** Most passengers had no parents or children aboard. The IQR is zero, causing positive values to be flagged as outliers.

These potential outliers were retained because they may represent genuine passenger characteristics rather than incorrect data.

```python
numeric_columns = ["Age", "Fare", "SibSp", "Parch"]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for col, ax in zip(numeric_columns, axes.flatten()):
    sns.histplot(data=df_clean, x=col, kde=True, ax=ax)
    ax.set_title(f"Distribution of {col}")
    ax.set_xlabel(col)
    ax.set_ylabel("Number of Passengers")

plt.tight_layout()
plt.savefig("feature_distributions.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Numerical Feature Distribution Analysis

Histograms were used to examine the distributions of Age, Fare, SibSp, and Parch.

**Age:** Most passengers were young adults, with a concentration around 25–30 years. The distribution extends from infants to older passengers. The peak around age 28 may partly reflect median imputation of missing values.

**Fare:** The distribution is strongly right-skewed. Most passengers paid relatively low fares, while a small number paid very high fares.

**SibSp:** Most passengers travelled without siblings or spouses. The frequency decreases as the number of siblings or spouses increases.

**Parch:** Most passengers had no parents or children aboard. Only a smaller proportion travelled with one or more parents or children.

Overall, the numerical features have different distribution patterns. Fare, SibSp, and Parch show strong right-skewness, while Age is concentrated among young adults.

### Categorical Feature Distribution Analysis

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_clean,
    x="Sex",
    hue="Survived"
)

plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.tight_layout()
plt.savefig("survival_by_gender.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
survival_by_gender = pd.crosstab(
    df_clean["Sex"],
    df_clean["Survived"],
    normalize="index"
) * 100

survival_by_gender.columns = ["Did Not Survive (%)", "Survived (%)"]

survival_by_gender
```

```python
survival_by_class = pd.crosstab(
    df_clean["Pclass"],
    df_clean["Survived"],
    normalize="index"
) * 100

survival_by_class.columns = ["Did Not Survive (%)", "Survived (%)"]

survival_by_class
```

### Survival Analysis by Passenger Class

The survival percentages were calculated separately for each passenger class.

- First class: 62.96% survived, while 37.04% did not survive.
- Second class: 47.28% survived, while 52.72% did not survive.
- Third class: 24.24% survived, while 75.76% did not survive.

First-class passengers had the highest survival rate, whereas third-class passengers had the lowest.

The results indicate a strong association between passenger class and survival outcomes. Survival rates decreased from first class to third class.

This pattern suggests that passenger class was an important factor associated with survival during the Titanic disaster.

```python
# Select numerical columns
correlation = df_clean.select_dtypes(include="number").corr()

# Create heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Titanic Dataset")
plt.tight_layout()

plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Correlation Heatmap Analysis

The correlation heatmap was used to examine linear relationships between numerical features.

Key findings:

- Survived and Pclass have a negative correlation of approximately -0.34. Higher passenger-class numbers, representing lower classes, are associated with lower survival.
- Survived and Fare have a positive correlation of approximately 0.26, suggesting that passengers who paid higher fares tended to have better survival outcomes.
- Survived and HasCabin have a positive correlation of approximately 0.32. Recorded cabin information is associated with higher survival.
- Pclass and Fare have a negative correlation of approximately -0.55, indicating that lower-numbered passenger classes generally had higher fares.
- Pclass and HasCabin have a negative correlation of approximately -0.73, suggesting that cabin information was more frequently recorded for passengers in higher classes.
- SibSp and Parch have a positive correlation of approximately 0.41, indicating an association between the numbers of siblings/spouses and parents/children aboard.

These relationships are associations rather than proof of causation. Correlations involving HasCabin may partly reflect differences in passenger class and the completeness of cabin records. PassengerId has little analytical value as an identifier.

```python
pairplot_data = df_clean[["Age", "Fare", "Pclass", "Survived"]].copy()

sns.pairplot(
    pairplot_data,
    hue="Survived",
    diag_kind="hist",
    corner=True,
    plot_kws={"alpha": 0.5, "s": 20}
)

plt.savefig("pairwise_scatter_plots.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Pairwise Scatter Plot Analysis

A pairplot was created to investigate relationships between Age, Fare, and Pclass, with points coloured according to survival status.

**Age Distribution:** Most passengers were between approximately 20 and 40 years old. Both survivors and non-survivors appear across different age groups.

**Fare Distribution:** Most passengers paid relatively low fares, while a small number paid much higher fares. The distribution is strongly right-skewed.

**Pclass Distribution:** Third class contains the largest number of passengers. First-class passengers show a higher proportion of survivors compared with third-class passengers.

**Age vs. Fare:** Most passengers are concentrated below a fare of 100 across different age groups. No strong linear relationship is visually apparent.

**Age vs. Pclass:** Passengers from different age groups are present in all three passenger classes.

**Fare vs. Pclass:** First-class passengers generally paid higher fares, while third-class passengers mostly paid lower fares.

Overall, the pairplot supports the earlier findings that passenger class and fare are associated with survival outcomes. However, visual associations do not establish causation.

### Observation: Survival by Embarkation Port

The survival rate varied across the three embarkation ports. Cherbourg (C) had the highest survival rate at **55.36%**, followed by Queenstown (Q) at **38.96%** and Southampton (S) at **33.90%**.

Southampton had the largest number of passengers, but non-survivors significantly outnumbered survivors. Cherbourg was the only port where survivors outnumbered non-survivors.

These results show an association between embarkation port and survival outcomes. However, embarkation port should not be considered a direct cause of survival because other factors, such as passenger class, gender, and fare, may also influence the results.

```python
readme_content = """
# Exploratory Data Analysis (EDA) - Titanic Dataset

## 1. Project Overview

This project performs Exploratory Data Analysis (EDA) on the Titanic
passenger dataset. The objective is to understand passenger information,
identify missing values, clean the data, detect outliers, and investigate
factors associated with passenger survival.

## 2. Dataset Information

- Dataset: Titanic Dataset
- Source: Kaggle Titanic Competition
- File analyzed: train.csv
- Total records: 891
- Original columns: 12
- Target variable: Survived

### Dataset Features

- PassengerId: Unique passenger identifier
- Survived: Survival status (0 = No, 1 = Yes)
- Pclass: Passenger class
- Name: Passenger name
- Sex: Passenger gender
- Age: Passenger age
- SibSp: Number of siblings/spouses aboard
- Parch: Number of parents/children aboard
- Ticket: Ticket number
- Fare: Ticket fare
- Cabin: Cabin information
- Embarked: Port of embarkation

## 3. Technologies Used

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Jupyter Notebook

## 4. Data Cleaning and Preprocessing

The following operations were performed:

1. Checked dataset structure, data types, and missing values.
2. Filled missing Age values using the median (28).
3. Filled missing Embarked values using the mode (S).
4. Created a HasCabin feature to indicate whether cabin information
   was available.
5. Removed the original Cabin column because approximately 77.1%
   of its values were missing.
6. Checked for duplicate records. No duplicate rows were found.
7. Detected potential outliers using the Interquartile Range (IQR)
   method.
8. Retained potential outliers because extreme values may represent
   genuine passenger characteristics rather than data errors.

## 5. Exploratory Data Analysis

The following analyses and visualizations were performed:

1. Numerical feature distributions
2. Categorical feature distributions
3. Boxplots for outlier detection
4. Correlation heatmap
5. Pairwise scatter plots
6. Survival count by gender
7. Survival count by embarkation port

All seven high-resolution visualization images are saved in the
visualizations folder.

## 6. Key Findings

### Gender and Survival

- Female survival rate: 74.20%
- Male survival rate: 18.89%

Female passengers had a substantially higher survival rate than
male passengers in this dataset.

### Passenger Class and Survival

- First class survival rate: 62.96%
- Second class survival rate: 47.28%
- Third class survival rate: 24.24%

Survival rates decreased from first class to third class.

### Embarkation Port and Survival

- Cherbourg (C): 55.36%
- Queenstown (Q): 38.96%
- Southampton (S): 33.90%

Cherbourg had the highest survival percentage, while Southampton
had the lowest.

### Correlation Analysis

Important observed correlations include:

- Survived and Pclass: approximately -0.34
- Survived and Fare: approximately +0.26
- Survived and HasCabin: approximately +0.32
- Pclass and Fare: approximately -0.55

These correlations indicate statistical associations, not proof
of causation. PassengerId is an identifier and is not a meaningful
predictive feature by itself.

### Outlier Analysis

The IQR method identified potential outliers in Age, Fare, SibSp,
and Parch. These observations were retained because they may be
valid passenger records.

## 7. Conclusion

The analysis indicates that survival outcomes in the Titanic dataset
were associated with passenger gender, passenger class, fare, and
embarkation port.

Female passengers and first-class passengers had higher survival
rates than male and third-class passengers.

The project demonstrates an EDA workflow covering data ingestion,
profiling, cleaning, missing-value treatment, outlier detection,
statistical analysis, and visualization.

## 8. Project Structure

- Titanic_EDA.ipynb: Jupyter Notebook containing the EDA analysis.
- README.md: Project documentation and summary of findings.
- titanic.zip: Original downloaded dataset archive.
- titanic_data/: Extracted dataset files.
  - train.csv
  - test.csv
  - gender_submission.csv
- visualizations/: Folder containing seven high-resolution PNG charts.

## 9. Dataset Source

Kaggle Titanic Competition:

https://www.kaggle.com/competitions/titanic/data
"""

with open("README.md", "w", encoding="utf-8") as file:
    file.write(readme_content)

print("README.md updated successfully!")

with open("README.md", "r", encoding="utf-8") as file:
    print(file.read()[:500])
```
